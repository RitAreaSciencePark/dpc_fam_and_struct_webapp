**Goal**: Concatenate all dataframes related to DPCFam Standard MC properties into a single, consistent dataframe.

```python
# 0. Imports
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import data_profiling # updated version of pandas-profiling
from data_profiling import ProfileReport
```

# I. Concatenating DataFrames

```python
# 1. Paths to files
root_path = "/u/mdmc/enyanduk/internship_areasciencepark/Dataframes/DPCFam"

# List of files to merge
files = [
    "mclist.csv",
    "sequence_information.csv",
    "pfam_comparison.csv",
    "lcregions.csv",
    "ccregions.csv",
    "disregions.csv",
    "tmregions.csv"
]

# Load all files into a list of DataFrames
dfs = [pd.read_csv(os.path.join(root_path, f)) for f in files]
```

```python
# Merge all dataframes:
df = (
    pd.concat(
        [d.set_index("MCID") for d in dfs],
        axis=1,
        join="outer"  # keeps all MCIDs
    )
    .reset_index()
 )
```

```python
df.head()
```

```python
# Drop some useless columns
df = df.drop(columns=["DAC", "%DAC", "%DACF", "%DACFA", "fred", "fext","pfam_seqs"])
```

```python
# New dataframe
df.head()
```

```python
# Rearrange columns properly
df = df[[
    "MCID", "seed_size", "average_lenght", "std_average_length", "%LC", "%CC", "%DIS", "TM", "DA", "%DA", "AvOv", "LABEL"
    ]]
df.head()
```

```python
# We rename columns properly to avoid conflicts in PostgreSQL:
df = df.rename(columns={
    "MCID": "mcid",
    "seed_size": "size_uniref50",
    "average_lenght": "avg_len",
    "std_average_length": "std_avg_len",
    "%LC": "lc_percent",
    "%CC": "cc_percent",
    "%DIS": "dis_percent",
    "TM": "tm",
    "DA": "pfam_da",
    "%DA": "da_percent",
    "AvOv": "avg_ov_percent",
    "LABEL": "overlap_label"
})
df.head()
```

```python
# I am curious about the subdataframe where pfam_da = "UNK"
df_unk = df[df["pfam_da"] == "UNK"]
print(f"Number of rows with pfam_da = 'UNK': {len(df_unk)}")
df_unk.head()
```

```python
# Hypothesis: Every time pfam_da is "UNK", da_percent = avg_ov_percent = overlap_label = NaN
# Let's verify this hypothesis.
condition = (df_unk["pfam_da"] == "UNK") & (df_unk["da_percent"].isnull()) & (df_unk["avg_ov_percent"].isnull()) & (df_unk["overlap_label"].isnull())
if condition.all():
    print("Every time pfam_da is 'UNK', da_percent, avg_ov_percent, and overlap_label are NaN.")
else:
    print("The hypothesis is not true. There are cases where pfam_da is 'UNK' but da_percent, avg_ov_percent, or overlap_label are not NaN.")
```

**Conclusion**: Tested hypothesis: If `pfam_da` = "UNK", then `da_percent`, `avg_ov_percent`, and `overlap_label`  are NaN.

# Normalisation

```python
# We perform some meaningful transformations to make the data more interpretable:

# T1: Rewrite each ID in mcid column as MCID: e.g.: 1 -> MC1
df["mcid"] = "MC" + df["mcid"].astype(str)

# T2: The size_uniref50 column stores integer values. We convert them to the int type:
df["size_uniref50"] = df["size_uniref50"].astype(int)

# T3 : In pfam_da column, we replace "UNK" by "UNKNOWN"
df["pfam_da"] = df["pfam_da"].replace("UNK", "UNKNOWN")

# Verified Hypothesis : Every time pfam_da is "UNK", da_percent = avg_ov_percent = overlap_label = NaN.
# T4: Normalization - Whenever pfam_da is "UNKNOWN", we set da_percent = avg_ov_percent = 0 and overlap_label = "NONE"
# Handle the numerical zeros first
df.loc[df["pfam_da"] == "UNKNOWN", ["da_percent", "avg_ov_percent"]] = 0
# Capitalize existing strings FIRST, then fill NaNs with our specific "NONE"
df["overlap_label"] = df["overlap_label"].str.capitalize().fillna("NONE") 

# T5: Percentages are more interpretable when multiplied by 100 and rounded to 2 digits. 
percent_cols = ["lc_percent", "cc_percent", "dis_percent", "da_percent", "avg_ov_percent"]
df[percent_cols] = df[percent_cols].mul(100).round(2)

# T6: Round specific cols to 2 digits:
round_cols = ["avg_len", "std_avg_len", "tm"]
df[round_cols] = df[round_cols].round(2)

# Final dataframe after transformations:
df.head()
```

```python
# Tested Hypothesis: Every time pfam_da is "UNKNOWN", da_percent and avg_ov_percent are 0, and overlap_label is "NONE"
# 1. Isolate the rows we care about
subset_unknown = df[df["pfam_da"] == "UNKNOWN"]

# 2. Check if all values in those specific rows meet the criteria
valid_da = (subset_unknown["da_percent"] == 0).all()
valid_ov = (subset_unknown["avg_ov_percent"] == 0).all()
valid_label = (subset_unknown["overlap_label"] == "NONE").all()

if valid_da and valid_ov and valid_label:
    print("Hypothesis Re-verified: All UNKNOWN rows are correctly normalized!")
else:
    print("Verification failed.")
```

**Conclusion**: We will use the verified hypothesis, which reflects the reality of the original paper, to help us correct errors in **DPCFamB**.
