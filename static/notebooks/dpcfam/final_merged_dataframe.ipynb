**Goal**: Concatenate all dataframes related to DPCFam Standard MC properties into a single, consistent dataframe.

```python
# 0. Imports
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import data_profiling # updated version of pandas-profiling
from data_profiling import ProfileReport
```

# I. Concatenating DataFrames

```python
# 1. Paths to files
root_path = "/u/mdmc/enyanduk/internship_areasciencepark/Dataframes/DPCFam"

# List of files to merge
files = [
    "mclist.csv",
    "sequence_information.csv",
    "pfam_comparison.csv",
    "lcregions.csv",
    "ccregions.csv",
    "disregions.csv",
    "tmregions.csv"
]

# Load all files into a list of DataFrames
dfs = [pd.read_csv(os.path.join(root_path, f)) for f in files]
```

```python
# Merge all dataframes:
df = (
    pd.concat(
        [d.set_index("MCID") for d in dfs],
        axis=1,
        join="outer"  # keeps all MCIDs
    )
    .reset_index()
 )
```

```python
df.head()
```

```python
# Drop some useless columns
df = df.drop(columns=["DAC", "%DAC", "%DACF", "%DACFA", "fred", "fext","pfam_seqs"])
```

```python
# New dataframe
df.head()
```

```python
# Rearrange columns properly
df = df[[
    "MCID", "seed_size", "average_lenght", "std_average_length", "%LC", "%CC", "%DIS", "TM", "DA", "%DA", "AvOv", "LABEL"
    ]]
df.head()
```

```python
# We rename columns properly to avoid conflicts in PostgreSQL:
df = df.rename(columns={
    "MCID": "mcid",
    "seed_size": "size_uniref50",
    "average_lenght": "avg_len",
    "std_average_length": "std_avg_len",
    "%LC": "lc_percent",
    "%CC": "cc_percent",
    "%DIS": "dis_percent",
    "TM": "tm",
    "DA": "pfam_da",
    "%DA": "da_percent",
    "AvOv": "avg_ov_percent",
    "LABEL": "overlap_label"
})
df.head()
```

```python
# We perform some meaningful transformations to make the data more interpretable:

# T1: Rewrite each ID in mcid column as MCID: e.g.: 1 -> MC1
df["mcid"] = "MC" + df["mcid"].astype(str)

# T2: The size_uniref50 column stores integer values. We convert them to the int type:
df["size_uniref50"] = df["size_uniref50"].astype(int)

# T3: For each value != NaN in cols, multiply it by 100 and round it
percent_cols = ["lc_percent", "cc_percent", "dis_percent", "da_percent", "avg_ov_percent"]
df[percent_cols] = df[percent_cols].fillna(0).mul(100).round(2)

# T4: Round specific cols to 2 digits:
round_cols = ["avg_len", "std_avg_len", "tm"]
df[round_cols] = df[round_cols].round(2)

# T5: In column pfam_da: replace UNK by UNKNOWN, in overlap_label: replace NaN by NONE
df["pfam_da"] = df["pfam_da"].replace("UNK", "UNKNOWN")
df["overlap_label"] = df["overlap_label"].str.capitalize().fillna("NONE")
df.head()
```

# II. Pfam Annotation Check

```python
# Create a subdataframe df2 where pfam_da is "UNKNOWN"
df2 = df[df["pfam_da"] == "UNKNOWN"]
df2.head()
```

```python
# Check whether every time pfam_da is "UNKNOWN", da_percent and avg_ov_percent are 0, and overlap_label is "NONE"
condition = (df2["pfam_da"] == "UNKNOWN") & (df2["da_percent"] == 0) & (df2["avg_ov_percent"] == 0) & (df2["overlap_label"] == "NONE")
if condition.all():
    print("Every time pfam_da is 'UNKNOWN', da_percent and avg_ov_percent are 0, and overlap_label is 'NONE'.")
else:
    print("There are cases where pfam_da is 'UNKNOWN', but da_percent, avg_ov_percent, and overlap_label do not match the expected values.")
```

**Conclusion**: `We will use the condition verified above, which reflects the reality of the original paper, and will help us fix errors in DPCFamB.`
